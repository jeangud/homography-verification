# Pixel value through a yaw rotation
Reproduces the pixel values figure used in the project README: for a single GTSRB image, plot how one pixel's intensity changes as the camera's yaw rotates from 0° to 10°.

```python
import matplotlib.pyplot as plt
import numpy as np

from vnn.datasets import GTSRB
from vnn.transforms import HomographyYaw
from vnn.visualization import plot_transform
```

## Choose image and pixel

```python
dataset = GTSRB(size=64)

idx = 34
img = dataset[idx][0].numpy()

# Pixel of interest
i, j = 34, 47
```

## Build the yaw transform

Yaw rotation centered on the image, with focal length `f = 30 px`, sweeping
from 0° to 10°.

```python
h, w = img.shape[-2:]
x0, y0 = w // 2, h // 2

p_min = np.deg2rad(0)
p_max = np.deg2rad(10)

transform = HomographyYaw(f=30, xc=x0, yc=y0)
```

## Plot and save

```python
fig = plot_transform(img, transform, p_min, p_max, i, j, padding_value=0)
fig.savefig("pixel_values.pdf", format="pdf", bbox_inches="tight", pad_inches=0)
plt.show()
```
